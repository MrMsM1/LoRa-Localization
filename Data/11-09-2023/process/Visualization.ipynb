```python
import numpy as np
import matplotlib.pyplot as plt
import json
from collections import defaultdict
```

```python
with open('output_final1.json', 'r') as file:
    receiver1_data = json.load(file)
with open('output_final2.json', 'r') as file:
    receiver2_data = json.load(file)
with open('output_final3.json', 'r') as file:
    receiver3_data = json.load(file)
with open('output_final4.json', 'r') as file:
    receiver4_data = json.load(file)
```

```python
accumulated_rssi1 = defaultdict(list)
accumulated_rssi2 = defaultdict(list)
accumulated_rssi3 = defaultdict(list)
accumulated_rssi4 = defaultdict(list)

averaged_rssi1 = defaultdict(list)
averaged_rssi2 = defaultdict(list)
averaged_rssi3 = defaultdict(list)
averaged_rssi4 = defaultdict(list)

for point in receiver1_data:
    coordinates = (point['x'], point['y'])
    accumulated_rssi1[coordinates].append(point['rssi'])
for point in receiver2_data:
    coordinates = (point['x'], point['y'])
    accumulated_rssi2[coordinates].append(point['rssi'])
for point in receiver3_data:
    coordinates = (point['x'], point['y'])
    accumulated_rssi3[coordinates].append(point['rssi'])
for point in receiver4_data:
    coordinates = (point['x'], point['y'])
    accumulated_rssi4[coordinates].append(point['rssi'])
    
for coordinate, rssi_values in accumulated_rssi1.items():
    averaged_rssi1[coordinate] = sum(rssi_values) / len(rssi_values)
for coordinate, rssi_values in accumulated_rssi2.items():
    averaged_rssi2[coordinate] = sum(rssi_values) / len(rssi_values)
for coordinate, rssi_values in accumulated_rssi3.items():
    averaged_rssi3[coordinate] = sum(rssi_values) / len(rssi_values)
for coordinate, rssi_values in accumulated_rssi4.items():
    averaged_rssi4[coordinate] = sum(rssi_values) / len(rssi_values)
    
x_vals_receiver1, y_vals_receiver1 = zip(*averaged_rssi1.keys())
rssi_vals_receiver1 = list(averaged_rssi1.values())
x_vals_receiver2, y_vals_receiver2 = zip(*averaged_rssi2.keys())
rssi_vals_receiver2 = list(averaged_rssi2.values())
x_vals_receiver3, y_vals_receiver3 = zip(*averaged_rssi3.keys())
rssi_vals_receiver3 = list(averaged_rssi3.values())
x_vals_receiver4, y_vals_receiver4 = zip(*averaged_rssi4.keys())
rssi_vals_receiver4 = list(averaged_rssi4.values())

plt.figure(figsize=(16, 12))

plt.subplot(2, 2, 1)
plt.scatter(x_vals_receiver1, y_vals_receiver1, c=rssi_vals_receiver1, cmap='cividis', s=100, alpha=1)
plt.colorbar(label='Average RSSI Values')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Average RSSI Data for Receiver 1')
plt.grid(True)

plt.subplot(2, 2, 2)
plt.scatter(x_vals_receiver2, y_vals_receiver2, c=rssi_vals_receiver2, cmap='cividis', s=100, alpha=1)
plt.colorbar(label='Average RSSI Values')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Average RSSI Data for Receiver 2')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.scatter(x_vals_receiver3, y_vals_receiver3, c=rssi_vals_receiver3, cmap='cividis', s=100, alpha=1)
plt.colorbar(label='Average RSSI Values')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Average RSSI Data for Receiver 3')
plt.grid(True)

plt.subplot(2, 2, 4)
plt.scatter(x_vals_receiver4, y_vals_receiver4, c=rssi_vals_receiver4, cmap='cividis', s=100, alpha=1)
plt.colorbar(label='Average RSSI Values')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Average RSSI Data for Receiver 4')
plt.grid(True)

plt.show()
```
